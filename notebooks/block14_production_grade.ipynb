## Étape 3 — Audit des artefacts de production

### Partie A — Mise en pratique

Pour chaque artefact critique, on vérifie :

L'empreinte SHA-256 permettra de détecter une modification d'un artefact

```python
# ==========================================
# CRITICAL PRODUCTION ARTIFACTS
# ==========================================

critical_artifacts = {

    "FastAPI":
        PROJECT_DIR
        / "api"
        / "main.py",

    "Streamlit":
        PROJECT_DIR
        / "dashboard"
        / "app.py",

    "PostgreSQL":
        PROJECT_DIR
        / "src"
        / "database"
        / "postgres.py",

    "Block 7 final state":
        MODEL_DIR
        / "block7_final_state.joblib",

    "Block 11 final state":
        MODEL_DIR
        / "block11_final_state.joblib",

    "Block 12 final state":
        MODEL_DIR
        / "block12_final_state.joblib",

    "Block 13 final state":
        MODEL_DIR
        / "block13_final_state.joblib",
}
```

```python
# ==========================================
# SHA-256 FUNCTION
# ==========================================

import hashlib


def sha256_file(path):

    hasher = hashlib.sha256()

    with open(
        path,
        "rb",
    ) as file:

        for chunk in iter(
            lambda: file.read(
                1024 * 1024
            ),
            b"",
        ):

            hasher.update(
                chunk
            )

    return hasher.hexdigest()
```

```python
# ==========================================
# ARTIFACT AUDIT
# ==========================================

from datetime import datetime


artifact_rows = []


for name, path in critical_artifacts.items():

    exists = path.exists()

    artifact_rows.append({

        "artifact":
            name,

        "path":
            str(
                path.relative_to(
                    PROJECT_DIR
                )
            ),

        "exists":
            exists,

        "size_mb":
            (
                path.stat().st_size
                / 1024**2
            )
            if exists
            else np.nan,

        "modified":
            (
                datetime.fromtimestamp(
                    path.stat().st_mtime
                )
            )
            if exists
            else None,

        "sha256":
            (
                sha256_file(
                    path
                )
            )
            if exists
            else None,
    })


production_artifacts = pd.DataFrame(
    artifact_rows
)


display(
    production_artifacts
)
```

```python
# ==========================================
# ARTIFACT VALIDATION
# ==========================================

assert (
    production_artifacts[
        "exists"
    ].all()
)


assert (
    production_artifacts[
        "sha256"
    ].notna().all()
)


print(
    "Artefacts critiques : OK"
)
```

Le test final doit rester statistiquement indépendant des décisions prises
pendant le développement.

l'architecture, les hyperparamètres ou les règles d'ensemble, le jeu de
test devient implicitement un jeu de validation.

On cherche au contraire à conserver :

$$
D_{\mathrm{train}},
D_{\mathrm{validation}}
\perp
D_{\mathrm{test\ final}}
$$

au niveau du processus de sélection du modèle.

Dans ce projet, l'année 2025 a été utilisée pour comparer et sélectionner
les différentes configurations.

L'année 2026 est réservée à une seule évaluation finale après verrouillage.

- l'environnement logiciel ;
- les décisions d'architecture ;
- les empreintes cryptographiques des artefacts ;
- la preuve que 2026 n'a pas encore été utilisée.

Après ouverture de 2026, les résultats devront être rapportés tels quels.
Ils ne devront pas servir à sélectionner une nouvelle version du modèle
dans cette expérience.
