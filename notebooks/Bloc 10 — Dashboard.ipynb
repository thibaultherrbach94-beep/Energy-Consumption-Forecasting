# Bloc 10 — Interactive Energy Forecasting Interface
## Dashboard Streamlit, visualisation et exploration des prévisions

## Étape 1 — Initialisation de Streamlit

### Partie C — Analyse et améliorations guidées par la théorie

Le dashboard affiche désormais simultanément la consommation réelle

$$
y_t
$$

et la consommation prédite

$$
\hat y_t.
$$

L'erreur ponctuelle est définie par :

$$
e_t = y_t - \hat y_t.
$$

Les performances peuvent être recalculées dynamiquement sur la
période sélectionnée.

Le RMSE est donné par :

$$
RMSE =
\sqrt{
\frac{1}{n}
\sum_{t=1}^{n}
e_t^2
}.
$$

Le MAE est :

$$
MAE =
\frac{1}{n}
\sum_{t=1}^{n}
|e_t|.
$$

Le biais moyen est :

$$
Bias =
\frac{1}{n}
\sum_{t=1}^{n}
e_t.
$$

Contrairement aux métriques globales de 2025, ces métriques locales
permettent d'identifier les périodes où le modèle fonctionne mieux
ou moins bien.

La visualisation temporelle des erreurs permet également de détecter
des structures non aléatoires qui pourraient être liées à des effets
calendaires, météorologiques ou événementiels.

Conclusion :

Le dashboard permet désormais de comparer visuellement les valeurs
réelles et prédites, puis de recalculer les principales métriques sur
la période choisie.

Cette analyse locale complète l'évaluation globale du modèle et prépare
l'étude détaillée des erreurs et des périodes difficiles.

## Étape 4 — Analyse détaillée des erreurs

### Partie C — Analyse et améliorations guidées par la théorie

L'analyse des erreurs permet de dépasser les métriques globales
comme le RMSE et le MAE.

Pour chaque instant, l'erreur est définie par :

$$
e_t = y_t - \hat y_t.
$$

L'étude de sa distribution permet de vérifier si les erreurs sont
centrées autour de zéro ou si un biais directionnel apparaît.

Les quantiles de l'erreur absolue :

$$
|e_t|
$$

permettent de mesurer le comportement de la queue de distribution.

Par exemple, le quantile à 95 % :

$$
Q_{0.95}(|e_t|)
$$

représente un seuil tel qu'environ 95 % des erreurs absolues sont
inférieures à cette valeur.

Les observations vérifiant :

$$
|e_t| > Q_{0.95}(|e_t|)
$$

peuvent être considérées comme des erreurs particulièrement fortes.

L'identification de ces observations permet ensuite de rechercher des
causes communes : événements calendaires, changements d'heure,
température extrême ou variations rapides de consommation.

Conclusion :

Le dashboard fournit désormais une analyse détaillée des erreurs :
distribution, quantiles, direction du biais, plus fortes erreurs et
détection des observations atypiques.

Cette étape permet de passer d'une évaluation moyenne du modèle à une
analyse locale des situations où les prédictions deviennent plus
difficiles.

## Étape 6 — météo, calendrier et événements

### Partie C — Analyse et améliorations guidées par la théorie

Le dashboard intègre désormais des variables de contexte météorologique
et calendaire.

Pour chaque instant de consommation, la température utilisée correspond
à la dernière observation météorologique disponible :

$$
t_{\mathrm{météo}}
\leq
t_{\mathrm{consommation}}.
$$

Cette contrainte évite d'utiliser une information future et préserve
la causalité temporelle du système.

Les erreurs du modèle peuvent désormais être analysées conditionnellement
à différentes variables, par exemple :

$$
E(|e_t|\mid H_t=h),
$$

où $H_t$ représente l'heure locale.

La comparaison entre semaine et week-end permet également d'étudier :

$$
E(|e_t|\mid Weekend=1)
$$

et :

$$
E(|e_t|\mid Weekend=0).
$$

La température apporte un contexte supplémentaire pour interpréter
les variations de demande électrique.

Cependant, une association entre température et consommation ne doit
pas être interprétée automatiquement comme un effet causal. Cette
distinction sera importante lorsque les méthodes causales seront
approfondies dans le projet.

Conclusion :

Le dashboard combine désormais les prévisions avec des informations
météorologiques et calendaires.

Il devient possible d'analyser simultanément la demande électrique,
la température, l'heure locale, les week-ends et les erreurs du modèle.

Cette contextualisation facilite l'identification des régimes où la
prévision devient plus difficile et prépare la comparaison interactive
de différentes périodes.

## Étape 7 — Comparer deux périodes

### Partie C — Analyse et améliorations guidées par la théorie

La comparaison de deux périodes permet d'étudier la stabilité du modèle
dans différents régimes de consommation.

Pour deux périodes $A$ et $B$, on peut notamment comparer :

$$
RMSE_A
\quad\text{et}\quad
RMSE_B,
$$

ainsi que :

$$
MAE_A,\qquad MAE_B
$$

et les niveaux moyens de consommation.

Cette analyse permet de déterminer si les performances du modèle
restent similaires dans différents contextes saisonniers.

L'ajout de la température permet également de contextualiser ces
différences.

Par exemple, une période froide peut présenter un niveau de demande
plus élevé qu'une période chaude.

Cependant, les différences observées entre deux périodes ne doivent
pas être interprétées automatiquement comme des effets causaux :
d'autres variables peuvent varier simultanément.

Conclusion :

Le dashboard permet désormais de comparer deux périodes de 2025 en
termes de consommation, pics de demande, température et performances
du modèle.

Cette fonctionnalité facilite l'étude de la stabilité du système de
prévision entre différents régimes saisonniers et différentes
conditions de consommation.

## Étape 8 — Robustesse, cache, tests et performances

### Partie C — Analyse et améliorations guidées par la théorie

La dernière étape du Bloc 10 améliore la robustesse et les performances
de l'interface.

Les appels coûteux aux endpoints de prédiction et de contexte sont
mis en cache pendant une durée limitée.

L'architecture devient :

$$
\text{Streamlit cache}
\rightarrow
\text{FastAPI}
\rightarrow
\text{PostgreSQL}.
$$

Pour une requête déjà calculée, le cache permet d'éviter un nouvel
accès inutile à l'API et à la base de données.

Les contrôles de santé de FastAPI et PostgreSQL restent volontairement
non mis en cache afin qu'ils représentent l'état courant du système.

Des tests automatiques sont également ajoutés pour vérifier
l'existence du dashboard, sa validité syntaxique et son exécution
sans exception.

L'ensemble du système peut ainsi être validé depuis les données
jusqu'à l'interface utilisateur.

Conclusion :

Le dashboard énergétique est désormais interactif, robuste et relié
aux différentes couches du projet.

Il permet d'explorer les prédictions, d'étudier les erreurs, de
contextualiser les résultats avec la météo et le calendrier, et de
comparer deux périodes de consommation.

La mise en cache améliore la réactivité de l'application tandis que
les tests automatisés permettent de vérifier son fonctionnement de
manière reproductible.
