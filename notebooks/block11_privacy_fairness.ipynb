## Étape 2 — Création du sandbox synthétique

### Partie A — Mise en pratique

Les données RTE du projet principal sont publiques et agrégées.

Pour étudier des problématiques qui nécessitent des observations
individuelles, on construit donc un dataset entièrement synthétique
représentant des foyers.

Ce dataset n'est pas utilisé pour remplacer les données énergétiques
réelles du projet.

Il sert uniquement aux expériences :

- d'équité algorithmique ;
- d'anonymisation ;
- de confidentialité différentielle lorsque celle-ci est activée.

Chaque foyer produit 48 observations correspondant à une journée
échantillonnée toutes les 30 minutes.

```python
# ==========================================
# SYNTHETIC SANDBOX PATH
# ==========================================

SYNTHETIC_PATH = (
    DATA_DIR
    / "block11_synthetic_households.joblib"
)


print(
    "Dataset synthétique :",
    SYNTHETIC_PATH
)
```

```python
# ==========================================
# SYNTHETIC DATA PARAMETERS
# ==========================================

N_HOUSEHOLDS = 1000

N_STEPS = 48

FREQUENCY = "30min"


print(
    "Nombre de foyers :",
    N_HOUSEHOLDS
)

print(
    "Observations par foyer :",
    N_STEPS
)

print(
    "Nombre total attendu :",
    N_HOUSEHOLDS * N_STEPS
)
```

```python
# ==========================================
# HOUSEHOLD TABLE
# ==========================================

household_ids = np.arange(
    N_HOUSEHOLDS
)


households = pd.DataFrame({

    "household_id":
        household_ids,

    "sensitive_group":
        rng.choice(
            [
                "A",
                "B",
            ],
            size=N_HOUSEHOLDS,
            p=[
                0.55,
                0.45,
            ],
        ),

    "household_size":
        rng.integers(
            1,
            6,
            size=N_HOUSEHOLDS,
        ),
})
```

```python
# ==========================================
# HOUSEHOLD CHARACTERISTICS
# ==========================================

households[
    "dwelling_type"
] = rng.choice(
    [
        "apartment",
        "house",
    ],
    size=N_HOUSEHOLDS,
    p=[
        0.60,
        0.40,
    ],
)


households[
    "electric_heating"
] = rng.binomial(
    n=1,
    p=0.45,
    size=N_HOUSEHOLDS,
)
```

```python
households[
    "energy_efficiency"
] = rng.choice(
    [
        "high",
        "medium",
        "low",
    ],
    size=N_HOUSEHOLDS,
    p=[
        0.25,
        0.50,
        0.25,
    ],
)
```

```python
# ==========================================
# SYNTHETIC GROUP EFFECT
# ==========================================

group_b_mask = (
    households[
        "sensitive_group"
    ]
    == "B"
)


n_group_b = int(
    group_b_mask.sum()
)


households.loc[
    group_b_mask,
    "energy_efficiency"
] = rng.choice(
    [
        "high",
        "medium",
        "low",
    ],
    size=n_group_b,
    p=[
        0.15,
        0.45,
        0.40,
    ],
)
```

```python
display(
    households.head()
)


print(
    "Shape households :",
    households.shape
)


display(
    households[
        "sensitive_group"
    ].value_counts(
        normalize=True
    )
)
```

```python
efficiency_by_group = pd.crosstab(
    households[
        "sensitive_group"
    ],
    households[
        "energy_efficiency"
    ],
    normalize="index",
)


display(
    efficiency_by_group
)
```

```python
# ==========================================
# TEMPORAL GRID
# ==========================================

timestamps = pd.date_range(
    start="2025-01-14 00:00:00",
    periods=N_STEPS,
    freq=FREQUENCY,
    tz="UTC",
)


time_df = pd.DataFrame({
    "timestamp_utc":
        timestamps
})


display(
    time_df.head()
)

print(
    "Nombre de timestamps :",
    len(time_df)
)
```

```python
# ==========================================
# HOUSEHOLD X TIME
# ==========================================

microdata = households.merge(
    time_df,
    how="cross",
)


print(
    "Shape microdata :",
    microdata.shape
)
```

```python
hour = (
    microdata[
        "timestamp_utc"
    ]
    .dt.hour
    .to_numpy()
    +
    microdata[
        "timestamp_utc"
    ]
    .dt.minute
    .to_numpy()
    / 60
)


microdata[
    "hour"
] = hour
```

```python
# ==========================================
# SYNTHETIC TEMPERATURE
# ==========================================

microdata[
    "temperature_c"
] = (
    2
    + 3
    * np.sin(
        2
        * np.pi
        * (
            microdata[
                "hour"
            ]
            - 14
        )
        / 24
    )
)
```

```python
# ==========================================
# DAILY CONSUMPTION PROFILE
# ==========================================

morning_peak = np.exp(
    -0.5
    * (
        (
            microdata[
                "hour"
            ]
            - 8
        )
        / 1.5
    ) ** 2
)


evening_peak = np.exp(
    -0.5
    * (
        (
            microdata[
                "hour"
            ]
            - 19
        )
        / 2.0
    ) ** 2
)


microdata[
    "daily_profile"
] = (
    0.70
    + 0.60 * morning_peak
    + 0.90 * evening_peak
)
```

```python
# ==========================================
# BASE CONSUMPTION
# ==========================================

base_consumption = (
    0.35
    + 0.18
    * microdata[
        "household_size"
    ]
)
```

```python
# ==========================================
# HEATING EFFECT
# ==========================================

heating_effect = (
    microdata[
        "electric_heating"
    ]
    * np.maximum(
        0,
        15
        - microdata[
            "temperature_c"
        ]
    )
    * 0.025
)
```

```python
# ==========================================
# DWELLING EFFECT
# ==========================================

dwelling_effect = np.where(
    microdata[
        "dwelling_type"
    ]
    == "house",
    0.30,
    0.0,
)
```

```python
# ==========================================
# ENERGY EFFICIENCY EFFECT
# ==========================================

efficiency_effect = (
    microdata[
        "energy_efficiency"
    ]
    .map({
        "high":
            -0.10,

        "medium":
            0.00,

        "low":
            0.18,
    })
    .to_numpy()
)
```

```python
# ==========================================
# RANDOM NOISE
# ==========================================

noise = rng.normal(
    loc=0.0,
    scale=0.12,
    size=len(
        microdata
    ),
)
```

```python
# ==========================================
# SYNTHETIC ENERGY CONSUMPTION
# ==========================================

microdata[
    "consumption_kwh"
] = (
    base_consumption
    * microdata[
        "daily_profile"
    ]
    + heating_effect
    + dwelling_effect
    + efficiency_effect
    + noise
)


microdata[
    "consumption_kwh"
] = (
    microdata[
        "consumption_kwh"
    ]
    .clip(
        lower=0.05
    )
)
```

```python
display(
    microdata.head()
)


display(
    microdata[
        "consumption_kwh"
    ].describe()
)
```

```python
observations_per_household = (
    microdata
    .groupby(
        "household_id"
    )
    .size()
)


display(
    observations_per_household
    .describe()
)
```

```python
assert (
    observations_per_household.min()
    == N_STEPS
)

assert (
    observations_per_household.max()
    == N_STEPS
)


print(
    "48 observations par foyer : OK"
)
```

```python
# ==========================================
# HOUSEHOLD-LEVEL FEATURES
# ==========================================

household_energy = (
    microdata
    .groupby(
        "household_id"
    )
    .agg(

        total_consumption_kwh=(
            "consumption_kwh",
            "sum"
        ),

        mean_consumption_kwh=(
            "consumption_kwh",
            "mean"
        ),

        peak_consumption_kwh=(
            "consumption_kwh",
            "max"
        ),

        mean_temperature_c=(
            "temperature_c",
            "mean"
        ),
    )
    .reset_index()
)
```

```python
household_dataset = (
    households
    .merge(
        household_energy,
        on="household_id",
        how="left",
    )
)
```

```python
# ==========================================
# CLASSIFICATION TARGET
# ==========================================

high_consumption_threshold = (
    household_dataset[
        "total_consumption_kwh"
    ]
    .quantile(
        0.75
    )
)


household_dataset[
    "high_consumption"
] = (
    household_dataset[
        "total_consumption_kwh"
    ]
    >= high_consumption_threshold
).astype(int)
```

```python
print(
    "Seuil :",
    high_consumption_threshold
)


display(
    household_dataset[
        "high_consumption"
    ].value_counts(
        normalize=True
    )
)
```

```python
# ==========================================
# TARGET BY SYNTHETIC GROUP
# ==========================================

target_by_group = (
    household_dataset
    .groupby(
        "sensitive_group"
    )
    .agg(

        n_households=(
            "household_id",
            "size"
        ),

        high_consumption_rate=(
            "high_consumption",
            "mean"
        ),

        mean_total_consumption=(
            "total_consumption_kwh",
            "mean"
        ),
    )
)


display(
    target_by_group
)
```

```python
# ==========================================
# DATA VALIDATION
# ==========================================

assert (
    len(
        households
    )
    == N_HOUSEHOLDS
)


assert (
    len(
        microdata
    )
    == (
        N_HOUSEHOLDS
        * N_STEPS
    )
)


assert (
    household_dataset[
        "household_id"
    ].nunique()
    == N_HOUSEHOLDS
)


assert (
    microdata[
        "consumption_kwh"
    ].isna().sum()
    == 0
)


assert (
    microdata[
        "consumption_kwh"
    ].min()
    > 0
)


assert (
    set(
        household_dataset[
            "high_consumption"
        ].unique()
    )
    <= {
        0,
        1,
    }
)


print(
    "Validation du sandbox : OK"
)
```

## Étape 3 — Anonymisation et \(k\)-anonymité

### Partie A — Mise en pratique

```python
# ==========================================
# LOAD BLOCK 11 SANDBOX
# ==========================================

if "household_dataset" not in globals():

    sandbox = joblib.load(
        SYNTHETIC_PATH
    )

    households = sandbox[
        "households"
    ]

    microdata = sandbox[
        "microdata"
    ]

    household_dataset = sandbox[
        "household_dataset"
    ]


print(
    "Foyers disponibles :",
    len(household_dataset)
)
```

```python
# ==========================================
# VARIABLE TYPES
# ==========================================

DIRECT_IDENTIFIERS = [
    "household_id",
]


QUASI_IDENTIFIERS = [
    "household_size",
    "dwelling_type",
    "electric_heating",
    "energy_efficiency",
]


SENSITIVE_ATTRIBUTES = [
    "sensitive_group",
]


TARGET_COLUMNS = [
    "high_consumption",
]


print(
    "Identifiants directs :",
    DIRECT_IDENTIFIERS
)

print(
    "Quasi-identifiants :",
    QUASI_IDENTIFIERS
)

print(
    "Attribut sensible :",
    SENSITIVE_ATTRIBUTES
)
```

```python
# ==========================================
# REMOVE DIRECT IDENTIFIER
# ==========================================

naive_anonymized = (
    household_dataset
    .drop(
        columns=DIRECT_IDENTIFIERS
    )
    .copy()
)


display(
    naive_anonymized.head()
)
```

### Partie C — Analyse et améliorations guidées par la théorie

La suppression d'un identifiant direct ne garantit pas à elle seule
l'anonymat.

Un individu peut parfois être distingué grâce à une combinaison de
quasi-identifiants.

Une classe d'équivalence regroupe les observations possédant les mêmes
valeurs de quasi-identifiants.

Le niveau de k-anonymité est défini par :

$$
k =
\min_g |G_g|,
$$

où $G_g$ représente une classe d'équivalence.

Un dataset est k-anonyme lorsque chaque combinaison de
quasi-identifiants correspond à au moins $k$ individus.

La généralisation réduit la précision de certaines variables afin
d'augmenter la taille des classes d'équivalence.

La suppression permet ensuite de retirer les observations appartenant
aux classes encore trop petites.

Cette méthode introduit cependant un compromis entre protection et
utilité :

$$
\text{Protection}\uparrow
\quad\Rightarrow\quad
\text{Précision des données}\downarrow.
$$

La pseudonymisation ne doit pas être confondue avec l'anonymisation.

Elle remplace un identifiant par un autre identifiant stable, mais les
observations correspondant au même individu peuvent encore être
reliées.

Enfin, la k-anonymité ne fournit pas une garantie probabiliste aussi
forte que la Differential Privacy.

Elle sera donc utilisée ici comme méthode introductive avant l'étude
des mécanismes de confidentialité différentielle.

Conclusion :

L'Étape 3 montre que supprimer un identifiant direct n'est pas
suffisant pour anonymiser un jeu de microdonnées.

Les quasi-identifiants peuvent créer des profils rares et permettre de
distinguer certains foyers.

La généralisation et la suppression permettent d'augmenter le niveau
de k-anonymité, au prix d'une perte d'information.

Ces transformations restent uniquement expérimentales dans ce projet.

Comme `PRIVACY_ENABLED = False`, le pipeline énergétique principal
continue d'utiliser les données publiques sans perturbation.

Aucune donnée de 2026 n'a été utilisée.

## Étape 4 — Differential Privacy

La confidentialité différentielle cherche à limiter l'influence d'une
unité individuelle sur le résultat publié par un algorithme.

Un mécanisme aléatoire $\mathcal{M}$ satisfait une garantie
$\varepsilon$-Differential Privacy si, pour deux datasets voisins
$D$ et $D'$ :

$$
P(\mathcal{M}(D)\in S)
\leq
e^\varepsilon
P(\mathcal{M}(D')\in S)
$$

pour tout ensemble de sorties possibles $S$.

Le paramètre :

$$
\varepsilon
$$

contrôle le compromis entre confidentialité et utilité.

Un petit $\varepsilon$ implique généralement davantage de bruit et une
protection plus forte.

Un grand $\varepsilon$ implique généralement moins de bruit et une
protection plus faible.

Dans ce projet, ces mécanismes sont étudiés uniquement dans le sandbox
synthétique.

Le pipeline énergétique principal conserve :

`PRIVACY_ENABLED = False`.

```python
# ==========================================
# PRIVATE QUERY — HIGH CONSUMPTION COUNT
# ==========================================

true_high_count = int(
    household_dataset[
        "high_consumption"
    ].sum()
)


print(
    "Nombre réel de foyers "
    "high consumption :",
    true_high_count
)
```

```python
# ==========================================
# GLOBAL SENSITIVITY
# ==========================================

COUNT_SENSITIVITY = 1.0


print(
    "Sensibilité du comptage :",
    COUNT_SENSITIVITY
)
```

```python
# ==========================================
# LAPLACE MECHANISM
# ==========================================

def laplace_mechanism(
    value,
    sensitivity,
    epsilon,
    random_generator,
):

    if epsilon <= 0:

        raise ValueError(
            "epsilon doit être > 0."
        )

    scale = (
        sensitivity
        / epsilon
    )

    noise = (
        random_generator
        .laplace(
            loc=0.0,
            scale=scale,
        )
    )

    return (
        value + noise
    )
```

```python
# ==========================================
# EPSILON GRID
# ==========================================

EPSILON_VALUES = [
    0.1,
    0.5,
    1.0,
    2.0,
    5.0,
    10.0,
]
```

```python
# ==========================================
# SINGLE DP RELEASE
# ==========================================

single_release_results = []


if RUN_PRIVACY_LAB:

    for epsilon in EPSILON_VALUES:

        private_count = (
            laplace_mechanism(
                value=true_high_count,
                sensitivity=COUNT_SENSITIVITY,
                epsilon=epsilon,
                random_generator=rng,
            )
        )


        single_release_results.append({

            "epsilon":
                epsilon,

            "true_count":
                true_high_count,

            "private_count":
                private_count,

            "absolute_error":
                abs(
                    private_count
                    - true_high_count
                ),

            "laplace_scale":
                (
                    COUNT_SENSITIVITY
                    / epsilon
                ),
        })


single_release_df = pd.DataFrame(
    single_release_results
)


display(
    single_release_df
)
```

```python
N_DP_SIMULATIONS = 500
```

```python
# ==========================================
# MONTE CARLO DP EXPERIMENT
# ==========================================

N_DP_SIMULATIONS = 500


dp_results = []


if RUN_PRIVACY_LAB:

    for epsilon in EPSILON_VALUES:

        scale = (
            COUNT_SENSITIVITY
            / epsilon
        )


        noises = rng.laplace(
            loc=0.0,
            scale=scale,
            size=N_DP_SIMULATIONS,
        )


        private_values = (
            true_high_count
            + noises
        )


        errors = (
            private_values
            - true_high_count
        )


        dp_results.append({

            "epsilon":
                epsilon,

            "scale":
                scale,

            "mean_private_count":
                np.mean(
                    private_values
                ),

            "bias":
                np.mean(
                    errors
                ),

            "mae":
                np.mean(
                    np.abs(
                        errors
                    )
                ),

            "rmse":
                np.sqrt(
                    np.mean(
                        errors ** 2
                    )
                ),

            "error_std":
                np.std(
                    errors
                ),
        })
```

```python
dp_results_df = pd.DataFrame(
    dp_results
)


display(
    dp_results_df
)
```

```python
dp_results_df[
    "theoretical_rmse"
] = (
    np.sqrt(2)
    * dp_results_df[
        "scale"
    ]
)


dp_results_df[
    "rmse_gap"
] = (
    dp_results_df[
        "rmse"
    ]
    - dp_results_df[
        "theoretical_rmse"
    ]
)


display(
    dp_results_df
)
```

```python
dp_results_df[
    "relative_mae_percent"
] = (
    100
    * dp_results_df[
        "mae"
    ]
    / true_high_count
)


display(
    dp_results_df[
        [
            "epsilon",
            "mae",
            "rmse",
            "theoretical_rmse",
            "relative_mae_percent",
        ]
    ]
)
```

```python
# ==========================================
# CONTRIBUTION CLIPPING
# ==========================================

CONSUMPTION_MIN = 0.0

CONSUMPTION_MAX = 100.0


clipped_total_consumption = (
    household_dataset[
        "total_consumption_kwh"
    ]
    .clip(
        lower=CONSUMPTION_MIN,
        upper=CONSUMPTION_MAX,
    )
)
```

```python
true_total_consumption = (
    clipped_total_consumption.sum()
)


SUM_SENSITIVITY = (
    CONSUMPTION_MAX
    - CONSUMPTION_MIN
)


print(
    "Consommation totale :",
    true_total_consumption
)

print(
    "Sensibilité :",
    SUM_SENSITIVITY
)
```

```python
# ==========================================
# PRIVATE TOTAL CONSUMPTION
# ==========================================

private_sum_results = []


if RUN_PRIVACY_LAB:

    for epsilon in EPSILON_VALUES:

        private_total = (
            laplace_mechanism(
                value=true_total_consumption,
                sensitivity=SUM_SENSITIVITY,
                epsilon=epsilon,
                random_generator=rng,
            )
        )


        private_sum_results.append({

            "epsilon":
                epsilon,

            "true_total_kwh":
                true_total_consumption,

            "private_total_kwh":
                private_total,

            "absolute_error_kwh":
                abs(
                    private_total
                    - true_total_consumption
                ),

            "scale":
                (
                    SUM_SENSITIVITY
                    / epsilon
                ),
        })


private_sum_df = pd.DataFrame(
    private_sum_results
)


display(
    private_sum_df
)
```

### Importance du clipping

La sensibilité d'une somme dépend de la contribution maximale qu'un
utilisateur peut apporter.

Sans borne sur la consommation individuelle, la sensibilité globale
de la somme peut devenir non bornée.

On impose donc :

$$
0 \leq C_i \leq 100.
$$

La sensibilité de la somme devient alors :

$$
\Delta f = 100.
$$

Le clipping joue donc un rôle fondamental dans la calibration du bruit
de confidentialité différentielle.

Une borne trop grande augmente inutilement le bruit.

Une borne trop petite tronque fortement les données.

Il existe donc également un compromis entre :

$$
\text{clipping}
\quad\text{et}\quad
\text{utilité}.
$$

### Agrégation préalable au niveau utilisateur

La User-Level Differential Privacy ne signifie pas automatiquement que
toutes les sensibilités sont multipliées par le nombre de lignes par
utilisateur.

Si les contributions temporelles sont d'abord agrégées au niveau du
foyer, chaque utilisateur peut de nouveau ne contribuer qu'une seule
valeur.

Pour le comptage :

$$
f(D)=\sum_i Y_i,
$$

avec :

$$
Y_i\in\{0,1\},
$$

la sensibilité utilisateur reste :

$$
\Delta f=1.
$$

L'agrégation par utilisateur constitue donc une technique importante
pour limiter la sensibilité.

```python
# ==========================================
# PRIVACY BUDGET COMPOSITION
# ==========================================

TOTAL_EPSILON_BUDGET = 1.0


QUERY_COUNTS = [
    1,
    2,
    5,
    10,
    20,
]


composition_rows = []


for n_queries in QUERY_COUNTS:

    epsilon_per_query = (
        TOTAL_EPSILON_BUDGET
        / n_queries
    )

    scale_per_query = (
        COUNT_SENSITIVITY
        / epsilon_per_query
    )

    composition_rows.append({

        "n_queries":
            n_queries,

        "epsilon_total":
            TOTAL_EPSILON_BUDGET,

        "epsilon_per_query":
            epsilon_per_query,

        "laplace_scale":
            scale_per_query,
    })


composition_df = pd.DataFrame(
    composition_rows
)


display(
    composition_df
)
```

```python
# ==========================================
# COMPOSITION MONTE CARLO
# ==========================================

N_COMPOSITION_SIMULATIONS = 500


rng_step5 = np.random.default_rng(
    RANDOM_STATE + 5
)


composition_results = []


for n_queries in QUERY_COUNTS:

    epsilon_per_query = (
        TOTAL_EPSILON_BUDGET
        / n_queries
    )

    scale = (
        COUNT_SENSITIVITY
        / epsilon_per_query
    )


    simulation_errors = []


    for _ in range(
        N_COMPOSITION_SIMULATIONS
    ):

        noises = rng_step5.laplace(
            loc=0.0,
            scale=scale,
            size=n_queries,
        )

        mean_release = (
            true_high_count
            + np.mean(
                noises
            )
        )

        simulation_errors.append(
            mean_release
            - true_high_count
        )


    simulation_errors = np.array(
        simulation_errors
    )


    empirical_rmse = np.sqrt(
        np.mean(
            simulation_errors ** 2
        )
    )


    theoretical_rmse = (
        np.sqrt(2)
        * COUNT_SENSITIVITY
        * np.sqrt(
            n_queries
        )
        / TOTAL_EPSILON_BUDGET
    )


    composition_results.append({

        "n_queries":
            n_queries,

        "epsilon_per_query":
            epsilon_per_query,

        "empirical_rmse":
            empirical_rmse,

        "theoretical_rmse":
            theoretical_rmse,
    })
```

```python
composition_results_df = pd.DataFrame(
    composition_results
)


display(
    composition_results_df
)
```

```python
# ==========================================
# CLIPPING GRID
# ==========================================

CLIPPING_BOUNDS = [
    50.0,
    60.0,
    70.0,
    80.0,
    100.0,
]


CLIPPING_EPSILON = 1.0


true_unclipped_sum = (
    household_dataset[
        "total_consumption_kwh"
    ].sum()
)
```

```python
# ==========================================
# CLIPPING / PRIVACY TRADE-OFF
# ==========================================

clipping_results = []


for upper_bound in CLIPPING_BOUNDS:

    clipped_values = (
        household_dataset[
            "total_consumption_kwh"
        ]
        .clip(
            lower=0.0,
            upper=upper_bound,
        )
    )


    clipped_sum = (
        clipped_values.sum()
    )


    clipping_bias = (
        clipped_sum
        - true_unclipped_sum
    )


    sensitivity = (
        upper_bound
    )


    dp_rmse = (
        np.sqrt(2)
        * sensitivity
        / CLIPPING_EPSILON
    )


    total_expected_rmse = np.sqrt(
        clipping_bias ** 2
        + dp_rmse ** 2
    )


    clipped_share = (
        100
        * (
            household_dataset[
                "total_consumption_kwh"
            ]
            > upper_bound
        ).mean()
    )


    clipping_results.append({

        "upper_bound_kwh":
            upper_bound,

        "clipped_share_percent":
            clipped_share,

        "clipping_bias_kwh":
            clipping_bias,

        "dp_theoretical_rmse":
            dp_rmse,

        "combined_expected_rmse":
            total_expected_rmse,
    })
```

La définition de l'unité de confidentialité est particulièrement
importante pour des séries temporelles individuelles.

Lorsqu'un foyer fournit 48 observations, une protection Row-Level et
une protection User-Level ne correspondent pas au même modèle de
menace.

Pour un comptage réalisé directement sur les lignes :

$$
\Delta_{\text{row}} = 1,
$$

alors qu'un foyer complet peut contribuer jusqu'à :

$$
\Delta_{\text{user}} = 48.
$$

Une agrégation préalable au niveau utilisateur peut cependant réduire
fortement cette sensibilité.

Pour un indicateur binaire calculé une fois par foyer :

$$
Y_i\in\{0,1\},
$$

la sensibilité utilisateur du comptage redevient :

$$
\Delta=1.
$$

La composition constitue un deuxième élément fondamental.

Pour plusieurs mécanismes utilisant respectivement :

$$
\varepsilon_1,\ldots,\varepsilon_m,
$$

la composition séquentielle simple donne :

$$
\varepsilon_{\text{total}}
=
\sum_{j=1}^{m}\varepsilon_j.
$$

Les publications répétées consomment donc progressivement le budget de
confidentialité.

Enfin, le clipping introduit un compromis supplémentaire.

Une borne faible réduit la sensibilité du mécanisme mais peut créer un
biais important.

Une borne élevée conserve davantage d'information mais augmente le
bruit nécessaire à la confidentialité.

Ce compromis peut être représenté par :

$$
RMSE_{total}
=
\sqrt{
Bias_{clip}^2
+
2
\left(
\frac{B}{\varepsilon}
\right)^2
}.
$$

Ces résultats montrent que la confidentialité différentielle nécessite
de choisir conjointement :

- l'unité de confidentialité ;
- la granularité des données ;
- les bornes de contribution ;
- le nombre de publications ;
- le budget epsilon.

Conclusion :

L'Étape 5 montre que la protection au niveau utilisateur peut être
nettement plus exigeante qu'une protection ligne par ligne lorsque
chaque utilisateur possède plusieurs observations.

L'agrégation préalable au niveau foyer permet toutefois de réduire la
sensibilité de certaines statistiques.

La composition montre également qu'un budget de confidentialité ne peut
pas être réutilisé gratuitement pour un nombre arbitraire de requêtes.

Enfin, l'expérience sur le clipping met en évidence un compromis entre
biais statistique et quantité de bruit nécessaire.

\max_a

Une valeur proche de zéro indique des taux de sélection similaires.

Elle compare notamment :

$$
TPR_a
=
P(\hat Y=1\mid Y=1,A=a)
$$

et :

$$
FPR_a
=
P(\hat Y=1\mid Y=0,A=a).
$$

Ces critères mesurent des propriétés différentes et peuvent mener à

## Étape 7 — Mitigation des écarts entre groupes

### Partie A — Mise en pratique

La baseline obtenue à l'étape précédente présente déjà :

- une très forte performance globale ;
- de faibles écarts entre groupes ;
- une Equalized Odds Difference faible.

L'objectif de cette étape n'est donc pas de forcer artificiellement une
parité parfaite.

On cherche plutôt à étudier le compromis entre :

$$
\text{performance}
$$

et :

$$
\text{fairness}.
$$

La méthode utilisée est Exponentiated Gradient avec une contrainte
d'Equalized Odds.

```python
# ==========================================
# FAIRLEARN MITIGATION IMPORTS
# ==========================================

from fairlearn.reductions import (
    EqualizedOdds,
    ExponentiatedGradient,
)
```

### Pourquoi utiliser Equalized Odds ?

La baseline présente déjà un écart de Demographic Parity relativement
faible.

Comme les taux réels de classe positive diffèrent légèrement entre les
groupes synthétiques, forcer exactement la même proportion de décisions
positives pourrait modifier artificiellement les prédictions.

On choisit donc plutôt de contrôler les différences de taux d'erreur.

Equalized Odds compare notamment :

$$
TPR_A
\quad\text{et}\quad
TPR_B,
$$

ainsi que :

$$
FPR_A
\quad\text{et}\quad
FPR_B.
$$

Cette contrainte est donc adaptée à l'analyse de la qualité des
prédictions conditionnellement à la véritable classe.

```python
# ==========================================
# FAIRNESS CONSTRAINT GRID
# ==========================================

FAIRNESS_EPS_VALUES = [
    0.001,
    0.01,
    0.02,
    0.05,
]
```

```python
# ==========================================
# MITIGATION BASE ESTIMATOR
# ==========================================

mitigation_pipeline = Pipeline([

    (
        "preprocessing",
        preprocessor,
    ),

    (
        "classifier",
        LogisticRegression(
            max_iter=1000,
            random_state=RANDOM_STATE,
        ),
    ),
])
```

```python
# ==========================================
# TRAIN MITIGATED MODELS
# ==========================================

mitigation_results = []

mitigation_models = {}


for fairness_eps in FAIRNESS_EPS_VALUES:

    constraint = EqualizedOdds(
        difference_bound=
            fairness_eps
    )


    mitigated_model = (
        ExponentiatedGradient(

            estimator=
                mitigation_pipeline,

            constraints=
                constraint,

            sample_weight_name=
                "classifier__sample_weight",

            eps=
                fairness_eps,

            max_iter=
                50,
        )
    )


    start_time = (
        time.perf_counter()
    )


    mitigated_model.fit(

        X_train,

        y_train,

        sensitive_features=
            sensitive_train,
    )


    train_time = (
        time.perf_counter()
        - start_time
    )


    y_pred_mitigated = (
        mitigated_model.predict(
            X_test,
            random_state=
                RANDOM_STATE,
        )
    )


    mitigated_accuracy = (
        accuracy_score(
            y_test,
            y_pred_mitigated,
        )
    )


    mitigated_balanced_accuracy = (
        balanced_accuracy_score(
            y_test,
            y_pred_mitigated,
        )
    )


    mitigated_f1 = (
        f1_score(
            y_test,
            y_pred_mitigated,
            zero_division=0,
        )
    )


    mitigated_dp_difference = (
        demographic_parity_difference(

            y_test,

            y_pred_mitigated,

            sensitive_features=
                sensitive_test,
        )
    )


    mitigated_eo_difference = (
        equalized_odds_difference(

            y_test,

            y_pred_mitigated,

            sensitive_features=
                sensitive_test,
        )
    )


    mitigation_results.append({

        "fairness_epsilon":
            fairness_eps,

        "accuracy":
            mitigated_accuracy,

        "balanced_accuracy":
            mitigated_balanced_accuracy,

        "f1":
            mitigated_f1,

        "dp_difference":
            mitigated_dp_difference,

        "eo_difference":
            mitigated_eo_difference,

        "training_seconds":
            train_time,
    })


    mitigation_models[
        fairness_eps
    ] = mitigated_model
```

### Partie C — Analyse et améliorations guidées par la théorie

Le modèle baseline présente déjà une très forte performance et des
écarts faibles entre groupes.

La mitigation est donc étudiée comme un compromis et non comme une
correction automatiquement nécessaire.

Exponentiated Gradient transforme le problème de classification sous
contrainte d'équité en une suite de problèmes de classification
pondérés.

La contrainte utilisée ici est Equalized Odds.

Elle cherche à réduire les écarts de :

$$
TPR
$$

et :

$$
FPR
$$

entre groupes.

Une contrainte plus stricte peut réduire les disparités, mais elle peut
également diminuer la performance prédictive.

On étudie donc simultanément :

$$
\Delta Accuracy
$$

et :

$$
\Delta EO.
$$

Le but n'est pas de maximiser une métrique unique, mais d'identifier le
compromis entre précision et contrainte d'équité.

Dans ce sandbox, l'attribut sensible est utilisé uniquement pendant
l'entraînement de la méthode de mitigation.

Le modèle mitigé peut ensuite produire des prédictions sans recevoir
directement cet attribut sensible au moment de l'inférence.

Conclusion :

L'Étape 7 compare la baseline à plusieurs modèles entraînés sous
contrainte d'Equalized Odds.

Cette comparaison permet d'observer si une réduction supplémentaire
des écarts entre groupes justifie ou non la perte éventuelle de
performance prédictive.

Dans notre cas, la baseline possède déjà des écarts relativement
faibles, ce qui rend particulièrement importante l'analyse du compromis
plutôt que l'application automatique d'une mitigation.

principal reste désactivée et aucune donnée de 2026 n'est utilisée.
